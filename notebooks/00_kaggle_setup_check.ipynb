## 1 · Is everything switched on?

```python
import os, sys, platform, subprocess, socket
print("Python  :", sys.version.split()[0], "on", platform.platform())
print("Kaggle  :", os.path.exists("/kaggle/working"))

# GPU check (nvidia-smi exists only when an accelerator is on)
try:
    print(subprocess.run(["nvidia-smi", "-L"], capture_output=True, text=True, timeout=10).stdout or "GPU     : none")
except Exception:
    print("GPU     : none (CPU session)")

# Internet check
try:
    socket.create_connection(("storage.googleapis.com", 443), timeout=5).close()
    print("Internet: ON ✅")
except OSError:
    print("Internet: OFF ❌  → Settings > Internet > On (phone-verified accounts only)")
```

## 3 · A first model, no download needed
scikit-learn ships with a small handwritten-digits dataset (8×8 pixels), a mini version of MNIST.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X, y = load_digits(return_X_y=True)
fig, axes = plt.subplots(1, 8, figsize=(10, 1.6))
for ax, img, label in zip(axes, X, y):
    ax.imshow(img.reshape(8, 8), cmap="gray_r"); ax.set_title(label); ax.axis("off")
plt.show()

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=0)
clf = LogisticRegression(max_iter=2000).fit(X_tr, y_tr)
print(f"Test accuracy: {clf.score(X_te, y_te):.3f}")
```
